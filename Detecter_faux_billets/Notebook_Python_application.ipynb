```python
import joblib
import pandas as pd

# Charger le modèle sauvegardé (assurez-vous que le fichier 'random_forest_model.pkl' se trouve dans le répertoire courant)
model = joblib.load("random_forest_model.pkl")

# Charger le fichier CSV avec les bons paramètres
file_path = r"C:\Users\Willkommen\OneDrive\Docs privés\HERVE\FORMATION\DATA_ANALYST_OC\PROJET_12\billets_test.csv"
df = pd.read_csv(file_path, sep=",", encoding="utf-8")

# Extraire les caractéristiques utilisées pour l'entraînement (en excluant la colonne 'ID')
features_classif = ["diagonal", "height_left", "height_right", "margin_low", "margin_up", "length"]
X_new = df[features_classif]

# Faire les prédictions avec le modèle sauvegardé.
# Le modèle a été entraîné avec sklearn, donc il doit accepter X_new directement.
predictions = model.predict(X_new)

# Ajouter les prédictions au DataFrame new_data.
# Ici, 0 signifie "Faux billet" et 1 signifie "Vrai billet".
df["prediction"] = predictions

# Probabilité associée à la prédiction faite (classe 0 ou 1 selon prediction)
df["prediction_proba"] = model.predict_proba(X_new).max(axis=1)

# Si vous souhaitez afficher des labels textuels plutôt que 0/1, vous pouvez créer une colonne supplémentaire :
df["prediction_label"] = df["prediction"].map({0: "Faux billet", 1: "Vrai billet"})

# Afficher les premières lignes du DataFrame avec les prédictions ajoutées
print(df.head())
```
